```python
# Import pandas for data handling
import pandas as pd

# Import scaling techniques from sklearn
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
# Load dataset into dataframe
df = pd.read_csv('fitness_activity_100+_preprocess.csv')

# Display first 5 rows
df.head()
```

```python
# Check number of rows and columns
print("Dataset Shape:")
print(df.shape)
```

```python
# Display dataset information
print(df.info())
```

```python
# Display statistical summary
print(df.describe())
```

```python
# Select only numerical columns
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

# Display numerical columns
print("Numerical Columns:")
print(numerical_columns)
```

```python
# Display original numerical data
print("Original Numerical Data:")
print(df[numerical_columns].head())
```

```python
# Create StandardScaler object
standard_scaler = StandardScaler()

# Apply standard scaling
standard_scaled_data = standard_scaler.fit_transform(df[numerical_columns])

# Convert scaled data into dataframe
standard_scaled_df = pd.DataFrame(
    standard_scaled_data,
    columns=numerical_columns
)

# Display scaled data
print("Standard Scaled Data:")
print(standard_scaled_df.head())
```

```python
# Check mean of scaled data
print("Mean after Standard Scaling:")
print(standard_scaled_df.mean())

# Check standard deviation
print("\nStandard Deviation after Standard Scaling:")
print(standard_scaled_df.std())
```

```python
# Create MinMaxScaler object
minmax_scaler = MinMaxScaler()

# Apply Min-Max scaling
minmax_scaled_data = minmax_scaler.fit_transform(df[numerical_columns])

# Convert scaled data into dataframe
minmax_scaled_df = pd.DataFrame(
    minmax_scaled_data,
    columns=numerical_columns
)

# Display scaled data
print("Min-Max Scaled Data:")
print(minmax_scaled_df.head())
```

```python
# Check minimum values
print("Minimum Values:")
print(minmax_scaled_df.min())

# Check maximum values
print("\nMaximum Values:")
print(minmax_scaled_df.max())
```

```python
print("Original Data")
print(df[numerical_columns].head())

print("\nStandard Scaled Data")
print(standard_scaled_df.head())

print("\nMin-Max Scaled Data")
print(minmax_scaled_df.head())
```
